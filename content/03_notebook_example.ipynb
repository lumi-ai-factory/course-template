## Code cells

Code cells are converted into standard code blocks with syntax highlighting and a copy button:

````python
import torch


def describe_device() -> str:
    """Report the GPU this job landed on."""
    if not torch.cuda.is_available():
        return "no GPU visible: are you running on a login node?"
    return torch.cuda.get_device_name(0)


print(describe_device())
````

Cell outputs are excluded by design. Plots generated by code cells are also omitted. To display them, save the plot to a file, place it in `public/assets/`, and link to it manually.

## Configuration

Notebooks do not require front matter. If you do not add any, the template falls back to default behavior:
- **Title**: Derived from the notebook's first heading.
- **Sidebar order**: Determined by a leading number in the filename (e.g., `03_...`). Make sure to include it if you don't add front matter.

If you choose to add custom front matter, the title and nav order are taken from there instead of the headings or filename. To add optional custom front matter, insert a new cell at the very top of the notebook containing your settings:

```yaml
---
title: "Chapter 1-1: authentication"
parent: "Getting Started with LUMI"
nav_order: 1
---
```

## Subchapters

You can make subchapters within the notebook just like you would within a markdown file! 

Simply insert a Markdown cell containing front matter anywhere **except** the first cell (which is reserved for the 'main' page). Everything below that block becomes a nested subpage.
