## Markdown cells

Markdown cells support all standard template features like formatting, [links](/chapter1), tables, and callouts.

## Code cells

Code cells are converted into standard code blocks with syntax highlighting and a copy button:

````python
import torch


def describe_device() -> str:
    """Report the GPU this job landed on."""
    if not torch.cuda.is_available():
        return "no GPU visible: are you running on a login node?"
    return torch.cuda.get_device_name(0)


print(describe_device())
````

Cell outputs are excluded by design. Plots generated by code cells are also omitted. To display them, save the plot to a file, place it in `public/assets/`, and link to it manually.

## Configuration

Notebooks do not require front matter.
- **Title**: Derived from the notebook's first heading.
- **Sidebar order**: Determined by the leading number in the filename (e.g., `03_...`). Make sure to include it.

To add optional custom front matter, insert a new Markdown cell at the very top of the notebook containing the front matter:

```yaml
---
parent: "Chapter 3: Getting Started with LUMI"
nav_order: 4
---
```

Only the first cell is read this way. A front matter block further down the notebook starts a new subpage instead, exactly as it does in a `.md` file.
