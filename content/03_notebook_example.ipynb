# Chapter 2: Jupyter Notebooks

This page is not a `.md` file. It is `content/03_notebook_example.ipynb`, an ordinary Jupyter notebook, shown exactly as the template renders it.

Drop a notebook into `content/` and it becomes a page like any other: it appears in the sidebar, gets its own address, joins the Previous and Next links at the bottom, and its headings fill the table of contents on the right. The notebook stays the only copy of your material, so editing it in Jupyter and pushing the change is all there is to it.

## Markdown cells

A markdown cell is a section of the page, and everything the template offers in a `.md` file works here too: **bold**, [links to other pages](/chapter1), tables, and callouts.

> [!note] Written in a notebook
> This callout was typed into a markdown cell, not into a `.md` file.

## Code cells

A code cell becomes a real code block, coloured for the notebook's language and with a copy button in the corner:

````python
import torch


def describe_device() -> str:
    """Report the GPU this job landed on."""
    if not torch.cuda.is_available():
        return "no GPU visible: are you running on a login node?"
    return torch.cuda.get_device_name(0)


print(describe_device())
````

Cell outputs are left out. A page shows the code your students are meant to run, not the results of whichever run happened to be saved in the file.

## Naming and ordering

A notebook needs no front matter. Its **title** is its first heading, the one at the top of this page. A notebook without a heading is named by its file instead, with underscores as spaces.

Its **place in the sidebar** comes from a number at the start of the file name. This file is `03_notebook_example.ipynb`, so it sits third. A notebook with no number goes to the end.

That is enough for most notebooks. To set anything else, add a cell at the very top, change its type from Code to Raw, and type the front matter into it:

```yaml
parent: "Chapter 1: Getting Started with LUMI"
description: "Running your first inference job."
```

`parent` nests this page under another one in the sidebar, named by that page's title, and `description` is the summary search engines show. A raw cell that is not front matter is left alone and shown as page content.

## Pictures

A picture reaches a notebook in one of two ways, and the template handles both.

**Pasted into a markdown cell.** Jupyter stores the picture inside the `.ipynb` itself, so it travels with the notebook and there is nothing for you to do.

**Kept as a file beside the notebook**, written as `![caption](./images/plot.png)`. Move those image files into `public/assets/` and they work with no change to the notebook, whatever the folder they were in was called. The picture below is one of them:

![The LUMI AI Factory logo](./images/lumi-ai-factory-logo.png)

That line still reads `./images/lumi-ai-factory-logo.png` in the notebook, while the file itself sits in `public/assets/`. Only the file name has to match, and the build log names any picture that is not there.

A plot drawn by a code cell is neither case: it is a cell output, so it does not appear. If your readers need to see one, save it to a file, put that file in `public/assets/`, and link to it.
